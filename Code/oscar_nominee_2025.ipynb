```python
import pandas as pd

#Read and convert the csv file for all best picture nominees and winners from 200-2024
train_df = pd.read_csv('/content/drive/MyDrive/best_picture_2000_to_2024 - oscars_df.csv.csv')
train_df.head()
```

```python
#Rename columns in the training data to match the column names in the testing data and strip the whitespace
train_df = train_df.rename(columns={"Director(s)" : "Director", "IMDB Rating" : "IMDb Rating"})
train_df.columns = train_df.columns.str.strip()
train_df.columns = train_df.columns.str.replace(' ', '')
train_df.head()
```

```python
import pandas as pd

#Read and convert the csv file for all eligble movies from 2025
test_df = pd.read_csv('/content/drive/MyDrive/american_films_2025 - american_films_2025.csv (2).csv')
test_df.head()
```

```python
#Rename columns in the test data to match the column names in the training data and strip whitespace
test_df = test_df.rename(columns={"Production company" : "Production Company", "Domestic gross" : "Domestic Income"})
test_df.columns = test_df.columns.str.strip()
test_df.columns = test_df.columns.str.replace(' ', '')
test_df.head()
```

Start data preprocessing

```python
#Since the Opening month feature is being done with integer encoding, and every month is assigned a static number, this can be done before splitting the dataset
month_mapping = {'Janurary': 1, 'Feburary': 2, 'March': 3,
                 'April': 4, 'May': 5, 'June': 6, 'July': 7,
                 'August': 8, 'September': 9, 'October': 10,
                 'November': 11, 'Novemeber' : 11, 'December': 12}

#Map the months to their respective number
train_df['opening_month_encoded'] = train_df['OpeningMonth'].astype(str).str.strip().map(month_mapping)
test_df['opening_month_encoded'] = test_df['OpeningMonth'].astype(str).str.strip().map(month_mapping)

#Drop the original 'Opening Month' column
train_df = train_df.drop(columns=['OpeningMonth'])
test_df = test_df.drop(columns=['OpeningMonth'])

print(train_df[['opening_month_encoded']].head())
print(test_df[['opening_month_encoded']].head())
```

```python
#Define features vs targets
X_train = train_df.drop(columns=['Title', 'Award'])
y_train = train_df['Award']

y_test = test_df['Award'].map({'Nominee' : 1, 'Winner' : 1, 'NaN' : 0})
X_test = test_df.drop(columns=['Title', 'Award'])
```

```python
#Split the training set into training and validation set
from sklearn.model_selection import train_test_split

X_train, X_validation, y_train, y_validation = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
```

Start feature scaling

```python
#Domestic gross income will be scaled with standard scaler
from sklearn.preprocessing import StandardScaler

X_train_scaled = X_train.copy()
X_validation_scaled = X_validation.copy()
X_test_scaled = X_test.copy()

#Fill empty blocks with 0
X_train_scaled['DomesticIncome'] = X_train_scaled['DomesticIncome'].fillna(0)
X_validation_scaled['DomesticIncome'] = X_validation_scaled['DomesticIncome'].fillna(0)
X_test_scaled['DomesticIncome'] = X_test_scaled['DomesticIncome'].fillna(0)

#Convert income from a string to numeric
X_train_scaled['DomesticIncome'] = pd.to_numeric(X_train_scaled['DomesticIncome'].astype(str).str.replace(',', '').str.replace('', ''))
X_validation_scaled['DomesticIncome'] = pd.to_numeric(X_validation_scaled['DomesticIncome'].astype(str).str.replace(',', '').str.replace('', ''))
X_test_scaled['DomesticIncome'] = pd.to_numeric(X_test_scaled['DomesticIncome'].astype(str).str.replace(',', '').str.replace('', ''))

#Fit the transform on the training data and apply to the validation and test set
scaler = StandardScaler()
X_train_scaled['DomesticIncome'] = scaler.fit_transform(X_train_scaled[['DomesticIncome']])
X_validation_scaled['DomesticIncome'] = scaler.transform(X_validation_scaled[['DomesticIncome']])
X_test_scaled['DomesticIncome'] = scaler.transform(X_test_scaled[['DomesticIncome']])

print(X_train_scaled[['DomesticIncome']].head())
```

```python
#Apply the same rules from the training to the validation set

#Because I cannot spell, match the typos to the correct genre in the training set
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bRomence\b', 'Romance', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bRomace\b', 'Romance', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bMusic\b', 'Musical', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bComdey\b', 'Comedy', regex=True)

#Combine the lesser appearing genres into one "other" category based on the genres that appeared in the training set
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r',\s*$', '', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'^\s*$', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bFamily\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bLegal\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bSuspense\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bHoliday\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bQuirky\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bSlapstick\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bConspiracy\b', 'Other', regex=True)
X_validation['MovieGenre'] = X_validation['MovieGenre'].str.replace(r'\bSteampunk\b', 'Other', regex=True)

#Check again for unique genres
unique_genres = X_validation['MovieGenre'].str.split(',').explode().unique()
print("The Unique Genres Are: ")
print(unique_genres)

genre_counts = X_validation['MovieGenre'].str.split(',').explode().value_counts()
print(genre_counts)
```

```python
#Apply the same rules from the training to the test set

#Because I cannot spell, match the typos to the correct genre in the training set
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bRomence\b', 'Romance', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bRomace\b', 'Romance', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bMusic\b', 'Musical', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bComdey\b', 'Comedy', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bThiller\b', 'Thriller', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bMuscial\b', 'Musical', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\Documentatry\b', 'Documentary', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\Sports\b', 'Sport', regex=True)

#Combine the lesser appearing genres into one "other" category based on the genres that appeared in the training set
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r',\s*$', '', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'^\s*$', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bFamily\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bLegal\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bSuspense\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bHoliday\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bQuirky\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bSlapstick\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bConspiracy\b', 'Other', regex=True)
X_test['MovieGenre'] = X_test['MovieGenre'].str.replace(r'\bSteampunk\b', 'Other', regex=True)

#Strip white space from the genre names
X_test['MovieGenre'] = X_test['MovieGenre'].apply(lambda x: [g.strip() for g in x.split(',') if g.strip()])

#Check again for unique genres
unique_genres = X_test['MovieGenre'].explode().unique()
print("The Unique Genres Are: ")
print(unique_genres)

genre_counts = X_test['MovieGenre'].explode().value_counts()
print(genre_counts)
```

```python
#Encode using multi-label binarizer
from sklearn.preprocessing import MultiLabelBinarizer

#Force the column into text strings
X_train['MovieGenre'] = X_train['MovieGenre'].astype(str).str.replace(r"[\[\]']", "", regex=True).apply(lambda x: [g.strip() for g in x.split(',') if g.strip()])
X_validation['MovieGenre'] = X_validation['MovieGenre'].astype(str).str.replace(r"[\[\]']", "", regex=True).apply(lambda x: [g.strip() for g in x.split(',') if g.strip()])
X_test['MovieGenre'] = X_test['MovieGenre'].astype(str).str.replace(r"[\[\]']", "", regex=True).apply(lambda x: [g.strip() for g in x.split(',') if g.strip()])

mlb = MultiLabelBinarizer()
X_train_encoded = mlb.fit_transform(X_train['MovieGenre'])
X_validation_encoded = mlb.transform(X_validation['MovieGenre'])
X_test_encoded = mlb.transform(X_test['MovieGenre'])

#Convert the lists back to dataframes
train_genres = pd.DataFrame(X_train_encoded, columns=mlb.classes_, index=X_train.index)
validation_genres = pd.DataFrame(X_validation_encoded, columns=mlb.classes_, index=X_validation.index)
test_genres = pd.DataFrame(X_test_encoded, columns=mlb.classes_, index=X_test.index)


X_train_scaled = pd.concat([X_train_scaled.drop(columns=['MovieGenre'], errors='ignore'), train_genres], axis=1)
X_validation_scaled = pd.concat([X_validation_scaled.drop(columns=['MovieGenre'], errors='ignore'), validation_genres], axis=1)
X_test_scaled = pd.concat([X_test_scaled.drop(columns=['MovieGenre'], errors='ignore'), test_genres], axis=1)

print(X_train_scaled.head())
```

```python
#Encode the directors
separator = ','

#Count to total frequency of each director in the training set
train_flatten = X_train_scaled['Director'].dropna().str.split(separator).explode()
freqs = train_flatten.value_counts(normalize=True)

#scale the rest of the data to the frequencies found above
X_train_scaled['Director'] = (
    X_train_scaled['Director'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

X_validation_scaled['Director'] = (
    X_validation_scaled['Director'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

X_test_scaled['Director'] = (
    X_test_scaled['Director'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

print(X_train_scaled.head())
```

```python
#Encode the Production Companies
separator = ','

train_flatten = X_train_scaled['ProductionCompany'].dropna().str.split(separator).explode()
freqs = train_flatten.value_counts(normalize=True)

X_train_scaled['ProductionCompany'] = (
    X_train_scaled['ProductionCompany'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

X_validation_scaled['ProductionCompany'] = (
    X_validation_scaled['ProductionCompany'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

X_test_scaled['ProductionCompany'] = (
    X_test_scaled['ProductionCompany'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

print(X_train_scaled.head())
```

```python
#Encode the stars
separator = ','

train_flatten = X_train_scaled['Stars'].dropna().str.split(separator).explode()
freqs = train_flatten.value_counts(normalize=True)

X_train_scaled['Stars'] = (
    X_train_scaled['Stars'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

X_validation_scaled['Stars'] = (
    X_validation_scaled['Stars'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

X_test_scaled['Stars'] = (
    X_test_scaled['Stars'].str.split(separator).explode().map(freqs).fillna(0).groupby(level=0).mean()
)

print(X_train_scaled.head())
```
