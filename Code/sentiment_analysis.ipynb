# Model Pipeline

The model pipeline combines two models. The first model will perform sentiment analysis on the tokenized quotes and output a sentiment. We will then use a second model that will take as input this sentiment and other relevant features in the dataset to classify the movie as an Oscar winner or not.

## Sentiment Analysis

For the purposes of sentiment analysis, only the following values are essential:

**Feature:** quote (string)

**Target:** rating (float)

### Tokenize Sentiments

Convert from String tokens into a numerical representation using TF-IDF.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

tokens = vectorizer.fit_transform(all_movies['quote'])

print(vectorizer.get_feature_names_out())
print(tokens)
```

### Convert Ratings into Integer Labels

Since the ratings can contain fractional values and our classes are purely integer values (ratings 1, 2, 3, 4, or 5), we round each non-integer rating down to the nearest integer.

```python
import numpy as np

print(f"ROW BEFORE CHANGES: \n{all_movies.iloc[:1]}")

# Update the y values by converting from fractional values into integers
# We ensure that any rating of 0.5 is converted into a 1, since that is the lowest possible value
# a class can have.
all_movies["sentiment_target"] = np.where(all_movies["rating"] < 1, 1, all_movies["rating"])

# We round down all other fractional values to the nearest possible integer.
all_movies["sentiment_target"] = all_movies["sentiment_target"].astype("int32")

print(f"\nROW AFTER CHANGES: \n{all_movies.iloc[:1]}")
```

### Create Train, Validation, and Test sets

```python
from sklearn.model_selection import train_test_split

X = tokens # X is the tokenized quotes
y = all_movies["sentiment_target"] # y is the integer ratings

# Define train and test sets using a 60/20/20 Train, Validation, and Test split.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=35, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=35, stratify=y_train)

print('X_train:', X_train.shape, 'X_val:', X_val.shape, 'X_test:', X_test.shape)
print('y_train:', y_train.shape, 'y_val:', y_val.shape, 'y_test:', y_test.shape)

X_train_svm = X_train
X_val_svm = X_val
X_test_svm = X_test

y_train_svm = y_train
y_val_svm = y_val
y_test_svm = y_test
```

### Training and Validation

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

train_accuracies_svm = []
val_accuracies_svm = []
precisions_svm = []
recalls_svm = []
f1s_svm = []
kernels = ['linear', 'rbf']
C_vals = [0.1, 1, 10]

# Iterate over the kernels and C values
for k in kernels:
  for c in C_vals:
    print(f"USING KERNEL {k} AND C = {c}:")

    # Train model, predict, and report acccuracy
    svm = SVC(C = c, kernel = k, random_state = 35)
    svm = svm.fit(X_train_svm, y_train_svm)
    train_prediction = svm.predict(X_train_svm)
    train_accuracy_svm = svm.score(X_train_svm, y_train_svm) # Calculate accuracy

    print(f"\tTraining Accuracy: {train_accuracy_svm}")

    # Perform validation on model
    val_prediction = svm.predict(X_val_svm)
    val_accuracy_svm = svm.score(X_val_svm, y_val_svm) # Calculate accuracy

    print(f"\tValidation Accuracy: {val_accuracy_svm}")

    # Calculate precision, recall, and f1
    precision_svm = precision_score(y_val_svm, val_prediction, average='weighted')
    recall_svm = recall_score(y_val_svm, val_prediction, average='weighted')
    f1_svm = f1_score(y_val_svm, val_prediction, average='weighted')

    train_accuracies_svm.append(train_accuracy_svm)
    val_accuracies_svm.append(val_accuracy_svm)
    precisions_svm.append(precision_svm)
    recalls_svm.append(recall_svm)
    f1s_svm.append(f1_svm)
```

When we run the training and validation loop above using kernels = ['linear', 'rbf']
and C_vals = [0.1, 1, 10], we got the following before the session timed out:

* USING KERNEL linear AND C = 0.1:
	- Training Accuracy: 0.568365888518342
	- Validation Accuracy: 0.5582188346203991
* USING KERNEL linear AND C = 1:
	- Training Accuracy: 0.7630701336476016
	- Validation Accuracy: 0.6104703640557617
* USING KERNEL linear AND C = 10:
	- Training Accuracy: 0.8951631102530027
	- Validation Accuracy: 0.5628322134189149
* USING KERNEL rbf AND C = 0.1:
	- Training Accuracy: 0.5040746220004514
	- Validation Accuracy: 0.5004513087955069
* USING KERNEL rbf AND C = 1:
	- Training Accuracy: 0.9098317494546275
	- Validation Accuracy: 0.6112726908033297
* USING KERNEL rbf AND C = 10:
	- Training Accuracy: 0.9957122439256789
	- Validation Accuracy: 0.6075619295958279


While using the RBF Kernel and C = 10 yielded the highest training accuracy, there was a large gap between the training and validation accuracy, indicating overfitting. So, we instead selected the Linear kernel with C = 1, since it had a smaller difference between the training and validation accuracy and a higher training accuracy than some other parameter combinations.
