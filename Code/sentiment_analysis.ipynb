# Model Pipeline

The model pipeline combines two models. The first model will perform sentiment analysis on the tokenized quotes and output a sentiment. We will then use a second model that will take as input this sentiment and other relevant features in the dataset to classify the movie as an Oscar winner or not.

## Sentiment Analysis

For the purposes of sentiment analysis, only the following values are essential:

**Feature:** quote (string)

**Target:** rating (float)

### Tokenize Sentiments

Convert From String Tokens into a Numerical Representation using TF-IDF.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

tokens = vectorizer.fit_transform(all_movies['quote'])

print(vectorizer.get_feature_names_out())
print(tokens)
```

### Convert Ratings into Integer Labels

Since the ratings can contain fractional values and our classes are purely integer values (ratings 1, 2, 3, 4, or 5), we round each non-integer rating down to the nearest integer.

```python
import numpy as np

print(f"ROW BEFORE CHANGES: \n{all_movies.iloc[:1]}")

# Update the y values by converting from fractional values into integers
# We ensure that any rating of 0.5 is converted into a 1, since that is the lowest possible value
# a class can have.
# We round down all other fractional values to the nearest possible integer.
all_movies["sentiment_target"] = np.where(all_movies["rating"] < 1, 1, all_movies["rating"])
all_movies["sentiment_target"] = all_movies["sentiment_target"].astype("int32")

print(f"\nROW AFTER CHANGES: \n{all_movies.iloc[:1]}")
```

### Create Train, Validation, and Test sets

```python
from sklearn.model_selection import train_test_split

X = tokens # X is the tokenized quotes
y = all_movies["sentiment_target"] # y is the integer ratings

# Define train and test sets using a 60/20/20 Train, Validation, and Test split.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=35, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=35, stratify=y_train)

print('X_train:', X_train.shape, 'X_val:', X_val.shape, 'X_test:', X_test.shape)
print('y_train:', y_train.shape, 'y_val:', y_val.shape, 'y_test:', y_test.shape)

X_train_svm = X_train
X_val_svm = X_val
X_test_svm = X_test

y_train_svm = y_train
y_val_svm = y_val
y_test_svm = y_test
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.model_selection import GridSearchCV

train_accuracies_svm = []
val_accuracies_svm = []
precisions_svm = []
recalls_svm = []
f1s_svm = []

k = 'linear'
c = 1

print(f"USING KERNEL {k} AND c = {c}:")

# Train model, predict, and report acccuracy
svm = SVC(C = c, kernel = k, random_state = 35)
svm = svm.fit(X_train_svm, y_train_svm)
train_prediction = svm.predict(X_train_svm)
train_accuracy_svm = svm.score(X_train_svm, y_train_svm) # Calculate accuracy

print(f"\tTraining Accuracy: {train_accuracy_svm}")

# Perform validation on model
val_prediction = svm.predict(X_val_svm)
val_accuracy_svm = svm.score(X_val_svm, y_val_svm) # Calculate accuracy

print(f"\tValidation Accuracy: {val_accuracy_svm}")

# Calculate precision, recall, and f1
precision_svm = precision_score(y_val_svm, val_prediction, average='weighted')
recall_svm = recall_score(y_val_svm, val_prediction, average='weighted')
f1_svm = f1_score(y_val_svm, val_prediction, average='weighted')

train_accuracies_svm.append(train_accuracy_svm)
val_accuracies_svm.append(val_accuracy_svm)
precisions_svm.append(precision_svm)
recalls_svm.append(recall_svm)
f1s_svm.append(f1_svm)
```

```python
logistic_movies = all_movies.copy()

# Convert columns with booleans into integer values: false = 0, true = 1
logistic_movies["isVerified"] = np.where(logistic_movies["isVerified"] == False, 0, 1)
logistic_movies["isSuperReviewer"] = np.where(logistic_movies["isSuperReviewer"] == False, 0, 1)
logistic_movies["hasSpoilers"] = np.where(logistic_movies["hasSpoilers"] == False, 0, 1)
logistic_movies["hasProfanity"] = np.where(logistic_movies["hasProfanity"] == False, 0, 1)

# Convert creationDate from "year-month-day" into year, month, and day by creating new columns
logistic_movies[['year', 'month', 'day']] = logistic_movies['creationDate'].str.split("-", expand=True)
logistic_movies = logistic_movies.drop(columns=["creationDate"])

# Encode movieTitle using an integer encoding, such that each unique movie title is assigned a unique number
# This is done by creating a dictionary of movie titles and number encodings.
# Since each dictionary movie title is given a new number, it is equivalent to integer encoding
title_dict = dict(enumerate(logistic_movies.movieTitle.unique()))

# Flip the keys and values of the title dictionary from "#: title" to "title: #"
title_dict = dict((v,k) for k,v in title_dict.items())

# Update the dataframe values using this dictionary
logistic_movies['movieTitle'] = logistic_movies['movieTitle'].replace(title_dict)

print(f"Sample rows:\n{logistic_movies[:1]}")
```

#### Create Train, Validation, and Test sets from Updated DF

```python
from sklearn.model_selection import train_test_split

# Drop irrelevant columns from X
X = logistic_movies.drop(columns=["oscarWinner", "quote", "sentiment_target", "rating", "userDisplayName", "userRealm", "reviewId", "userId"])

# Create the set y, which is a prediction of winner or not
y = logistic_movies["oscarWinner"]

# Define train and test sets using a 60/20/20 Train, Validation, and Test split.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=35, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=35, stratify=y_train)

print('X_train:', X_train.shape, 'X_val:', X_val.shape, 'X_test:', X_test.shape)
print('y_train:', y_train.shape, 'y_val:', y_val.shape, 'y_test:', y_test.shape)

X_train_lr = X_train
X_val_lr = X_val
X_test_lr = X_test

y_train_lr = y_train
y_val_lr = y_val
y_test_lr = y_test
```

#### Run Model

After testing a variety of penalty and regularization values, we got the following training and validation accuracies:


| Penalty | Regularization | Train Accuracy | Val Accuracy | Test Accuracy |
| :--- | :--- | :--- | :--- | :--- |
| L2 | 1 | 1.0| 1.0| 1.0|
| L2 | 0.9 |0.9996744791666666 | 1.0 |1.0 |

| L1 | 0.8 |0.9755859375 | 0.98046875 | 1.0|

Based on this, we can move forward with a penalty of L2 and a regularization of 1.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Test penalties and regularizations

# Create an instance of the logistic regression model with a random state of 35
logistic_model = LogisticRegression(penalty = 'l2', C = 1, random_state=35)

# Train model, predict, and report acccuracy
logistic_model.fit(X_train_lr, y_train_lr)
train_prediction = logistic_model.predict(X_train_lr)
train_accuracy_lr = accuracy_score(y_train_lr, train_prediction) # Calculate accuracy

print(f"Training Accuracy: {train_accuracy_lr}")

# Perform validation on model
val_prediction = logistic_model.predict(X_val_lr)
val_accuracy_lr = accuracy_score(y_val_lr, val_prediction) # Calculate accuracy

print(f"Validation Accuracy: {val_accuracy_lr}")
```

#### Report Metrics

```python
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from prettytable import PrettyTable


# Calculate precision, recall and f1 score
precision_lr = precision_score(y_val_lr, val_prediction)
recall_lr = recall_score(y_val_lr, val_prediction)
f1_lr = f1_score(y_val_lr, val_prediction)

# Print metrics
# Code created with assistance from this link: https://stackoverflow.com/questions/9535954/printing-lists-as-tabular-data
t = PrettyTable(['Metric', 'Value'])
t.add_row(['Train Accuracy', train_accuracy_lr])
t.add_row(['validation Accuracy', val_accuracy_lr])
t.add_row(['Precision', precision_lr])
t.add_row(['Recall', recall_lr])
t.add_row(['F1 Score', f1_lr])
print(t)
```

#### Test Model
